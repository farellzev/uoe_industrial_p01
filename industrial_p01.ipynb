```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from neumann_diffusion_1d import concentration, concentration_grid, mass
```

```python
D, d, N = 1.0, 1.0, 800     # diffusivity, half-domain, series truncation
xs = np.linspace(-d, d, 2001)

# Evaluate at a single time:
t = 0.05
c = concentration_grid(xs, t=t, D=D, d=d, N=N)

# Mass conservation check (≈1.0):
print(mass(xs, c))
```

```python
D = 1.0
d = 1.0
N = 800
xs = np.linspace(-d, d, 2001)
times = [0.0005, 0.005, 0.05, 0.5]

plt.figure()
for t in times:
    cs = concentration_grid(xs, t=t, D=D, d=d, N=N)
    plt.plot(xs, cs, label=f"t={t:g}")

plt.xlabel("x")
plt.ylabel("C(x, t)")
plt.title("1D Diffusion with Neumann BC on [-d,d] (D=1, d=1)")
plt.legend()
plt.show()
```
